## Now i will apply Stemming in the tag columns in new_df


```python
import nltk
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
ps = PorterStemmer()
```

```python
def stemming(text):
    Lt = []
    for i in text.split():
        Lt.append(ps.stem(i))
    return " ".join(Lt)
```

```python
new_df["tag"] = new_df["tag"].apply(stemming)
```

```python
new_df["tag"][0]
```

```python
#Lets convert into lower case
new_df["tag"] = new_df["tag"].apply(lambda x:x.lower())
```

```python
new_df.head(2)
```

```python
#Now we will do vectoization of tags

from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 3000, stop_words = "english")
# I am choosing less number of maximum features to avoid sparse matrix
```

```python
tag_vectors = cv.fit_transform(new_df["tag"]).toarray()
```

```python
tag_vectors.shape
```

```python
#Following are the most common words in the corpus
j = 0
for i in cv.get_feature_names_out():
    print(i)
    j+=1
    if j == 20:
        break
```

```python
#Now we will utilise cosine similarity
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(tag_vectors)
```

```python
similarity.shape
```

```python
similarity
```

```python
#Enumerate function to add a counter to a iterable
# Lambda x:x[1] to ensure that sorting occurs based on vector value and not index value

name = "Gandhi"
index = new_df[new_df["title"]==name].index[0]
sorted(enumerate(similarity[index]), reverse=True, key = lambda x:x[1])
#sorted(enumerate(similarity[index]), reverse = True, key = lambda:x:x[1])
```

The above list shows the index of the movie which are closest to movie Gandhi in decreasing order. Now we just have to convert it into a function to accept any movie

```python
# Final recommendation function
def recom(movie):
    index = new_df[new_df["title"]==movie].index[0]
    distances = sorted(enumerate(similarity[index]), reverse=True, key = lambda x:x[1])
    for i in distances[1:11]:
        print(new_df.iloc[i[0]].title)
```

```python
recom("Pirates of the Caribbean: On Stranger Tides")
```
